Objective: This report analyzes the secuirty risks of publicly exposed IP cameras using Shodan data.

```python
# Initial imports

import shodan
import os
from dotenv import load_dotenv
import json


load_dotenv()
SHODAN_API_KEY = os.getenv("SHODAN_API_KEY")

if not SHODAN_API_KEY:
     raise ValueError("Error: SHODAN_API_KEY not found. Please seit it up in the .env file.")
```

```python
# Initialize Shodan API client

api = shodan.Shodan(SHODAN_API_KEY)
```

```python
try:
    info = api.info()  # Get API account details
    print("✅ API Key is valid. Plan:", info)
except shodan.APIError as e:
    print("❌ API Error:", e)
```

```python
# Now we can define our query

query = 'port:554 OR port:443 OR "IP Camera"'

try:
    print("[+] Searching for IP cameras...")
    results = api.search(query, limit=50)

    # Save results securely
    os.makedirs("data", exist_ok=True)
    output_file="data/shodan_results.json"

    with open(output_file, "w") as file:
        json.dump(results, file, idnent=4)

    print(f"[+] Found {len(results['matches'])} results. data saved to '{output_file}'.")

except shodan.APIError as e:
    print(f"Shodan API error: {e}")
```
